```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service import jobs
```

```python
login = "mariiasirska"
cluster_id = "0828-065013-gln8ogbm"
notebook_path = "/Workspace/Users/sirskam123@softserve.academy/databricks_labs/lab3/eventhub_consumer.ipynb"
```

```python
w = WorkspaceClient()

job = w.jobs.create(
    name=f"{login}_lab3_eventhub_to_bronze",
    tasks=[
        jobs.Task(
            task_key="eventhub_to_bronze",
            existing_cluster_id=cluster_id,
            notebook_task=jobs.NotebookTask(notebook_path=notebook_path),
        )
    ],
    schedule=jobs.CronSchedule(
        quartz_cron_expression="0 0 * * * ?",  # hourly
        timezone_id="UTC",
        pause_status=jobs.PauseStatus.PAUSED,
    ),
    max_concurrent_runs=1,
)
print("job_id:", job.job_id)
```

Trigger one run manually:

```python
run = w.jobs.run_now(job_id=job.job_id).result()
print(run.state.result_state)
```
