```python
login = "mariiasirska"
catalog = "dbr_dev_ua5816bd"
silver = f"{catalog}.{login}_silver"
```

```python
tables = ["orders", "customers", "products", "order_items", "reviews", "sellers"]
```

```python
def st(name):
    return f"{silver}.olist_{name}_silver"
```

```python
for t in tables:
    spark.sql(f"""
    ALTER TABLE {st(t)} SET TBLPROPERTIES (
      'delta.autoOptimize.optimizeWrite' = 'true',
      'delta.autoOptimize.autoCompact' = 'true',
      'delta.deletedFileRetentionDuration' = 'interval 7 days',
      'delta.logRetentionDuration' = 'interval 30 days'
    )""")
```

```python
for t in tables:
    full = st(t)
    spark.sql(f"OPTIMIZE {full}")
    spark.sql(f"VACUUM {full}")
    files = spark.sql(f"DESCRIBE DETAIL {full}").first()["numFiles"]
    print(f"{t}: maintained, data files = {files}")
```
